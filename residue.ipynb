## 13. EDA takeaways

1. **Join**: `Client` and `Record` are a clean 1:1 merge on `Customer_ID` → 100k × 100 features.
2. **Target**: `churn` is balanced (~49.6% churn / 50.4% stay), so accuracy is usable.
3. **Missing data**: concentrated in Client demographics (`numbcars` ~49%, dwelling/owner/income 22–38%). Record usage fields are mostly complete (&lt;1%).
4. **Stronger single-feature signals** (still modest):
   - `eqpdays`: churners ~58 days older handsets on average; highest |corr| with churn (~0.11)
   - `hnd_price`: churners have cheaper handsets (negative corr ~−0.10)
   - `mou_Mean` / `change_mou`: churners use fewer minutes and show a larger recent usage drop
5. **Data quality**: a few negative `eqpdays`; `rev_Mean` / `totmrc_Mean` can be slightly negative; `change_*` has extreme tails — clip or winsorize before modeling.
6. **Redundancy**: MOU/revenue columns across windows are highly correlated — keep a lean feature set.
7. **Next steps (done in modeling notebook)**: encode categoricals (including explicit Missing), impute/drop sparse demographics, drop `Customer_ID`, train RF (primary) + logistic regression (counterpart).
